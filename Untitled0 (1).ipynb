1. append() = Adds one item to the end of the list

```python
fruits = ["apple", "banana"]
fruits.append("mango")
print(fruits)
```

2. extend() = adds multiple items to the end of a list

```python
fruits = ["apple", "banana"]
fruits.extend(["mango", "orange"])
print(fruits)
```

 reverse() = reverses the current order of the list

```python
fruits = ["apple", "banana", "mango"]
fruits.reverse()
print(fruits)
```

sort() = sorts the list in ascending order by default

```python
numbers = [5, 2 , 8, 1, 3]
numbers.sort()
print(numbers)
```

```python
# pop() = removes and returns an item from the list. Without an index pop() removes the last item
fruits = ["apple", "banana", "mango"]
removed = fruits.pop(1)
print(removed)
print(fruits)
```

```python
# remove() = removes the first matching item from a list
fruits = ["apple", "banana", "mango"]
fruits.remove("banana")
print(fruits)
```

```python
# insert() = adds an item at a specific index
fruits = ["apple", "banana", "orange"]
fruits.insert(2, "mango")
print(fruits)
```

```python
# count() = counts how many times an item appears in a list.
numbers = [1, 2, 3, 4, 2, 2, 4]
print(numbers.count(2))
```

```python
# index() = returns the index of the first matching item.
fruits = ["apple", "banana", "mango"]
position = fruits.index("banana")
print(position)
```

```python
# clear() = removes all items from a list
fruits = ["apple", "banana", "mango"]
fruits.clear()
print(fruits)
```
